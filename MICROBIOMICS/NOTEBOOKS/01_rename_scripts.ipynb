```python
read_count=pd.read_csv("../TABLES/read_count.csv", sep="\t")

def normalize_read_count_path(path):
    path = str(path).strip()
    if path.startswith("./"):
        return path
    marker = "/sequencing_runs/"
    if marker in path:
        return "./" + path.split(marker, 1)[1].lstrip("/")
    return "./" + path.lstrip("/")

read_count["Path"] = read_count["Path"].map(normalize_read_count_path)
read_count["file"]=read_count["Path"].str.split("/").str[-1]
read_count["folder"]=read_count["Path"].str.split("/").str[-2]
read_count["read"]=read_count["file"].str.split("_R._").str[0]
read_count["read_fixed"]=read_count["read"].str.upper()
read_count
```

```python
def list_files_with_prefix(folder_path, prefix):
    try:
        folder = str(folder_path).rstrip("/").split("/")[-1]
        relative_folder = f"./{folder}/"
        paths = read_count["Path"].astype(str)

        matching_files = read_count[
            paths.str.startswith(relative_folder) &
            paths.str.contains(prefix, regex=False)
        ]["Path"].tolist()

        matching_files.sort()
        return matching_files

    except Exception:
        return []
```

```python
# Merge two DataFrames (metadata_2020_washed_amplicon and df_folder_structure) 
# on the columns "run_ID" and "dir_name" respectively, keeping all rows from metadata_2020_washed_amplicon 
# (left join).

metadata_2020_washed_amplicon_full = metadata_2020_washed_amplicon.merge(
    df_folder_structure, left_on="run_ID", right_on="dir_name", how="left"
)

# Filter rows to keep only those where the "seq_sample_ID" column is not NaN.
metadata_2020_washed_amplicon_full = metadata_2020_washed_amplicon_full[
    metadata_2020_washed_amplicon_full["seq_sample_ID"].notna()
]

# Create a new column "suggested_correct_seq_sample_ID". If "correct_seq_sample_ID" is NaN,
# it will use the value from "seq_sample_ID", otherwise, it uses "correct_seq_sample_ID".
metadata_2020_washed_amplicon_full['suggested_correct_seq_sample_ID'] = \
    metadata_2020_washed_amplicon_full['correct_seq_sample_ID'].fillna(
        metadata_2020_washed_amplicon_full['seq_sample_ID']
    )

# Replace underscores with hyphens and remove the prefix "washed-1-" from the values in 
# the "suggested_correct_seq_sample_ID" column.
metadata_2020_washed_amplicon_full['suggested_correct_seq_sample_ID'] = metadata_2020_washed_amplicon_full['suggested_correct_seq_sample_ID'].str.replace("_", "-")
metadata_2020_washed_amplicon_full['suggested_correct_seq_sample_ID'] = metadata_2020_washed_amplicon_full['suggested_correct_seq_sample_ID'].str.replace("washed-1-", "")

# Sort the DataFrame by the "run_ID" column in ascending order.
metadata_2020_washed_amplicon_full = metadata_2020_washed_amplicon_full.sort_values("run_ID")

# End of code, the resulting DataFrame is assigned to metadata_2020_washed_amplicon_full.
metadata_2020_washed_amplicon_full
```

```python
# Iterate over each row and print files
metadata_2020_washed_amplicon_full["forward_fq"]=""
metadata_2020_washed_amplicon_full["reverse_fq"]=""
metadata_2020_washed_amplicon_full["current_seq_sample_ID"]=""
metadata_2020_washed_amplicon_full["corrected_forward_fq"]=""
metadata_2020_washed_amplicon_full["corrected_reverse_fq"]=""

total_rows = len(metadata_2020_washed_amplicon_full)  # Get the total number of rows
n=0
for index, row in metadata_2020_washed_amplicon_full.iterrows():
    try:
        folder_path = row['full_path']  # Get the folder path
        prefix = row['seq_sample_ID'] + "_"  # Get the prefix from the column
        matching_files = list_files_with_prefix(folder_path, prefix)
        correct_seq_sample_ID = row['suggested_correct_seq_sample_ID']

        # Check if the number of matching files is not equal to 2
        if len(matching_files) == 2:
            metadata_2020_washed_amplicon_full.loc[index, "forward_fq"] = matching_files[0]
            metadata_2020_washed_amplicon_full.loc[index, "reverse_fq"] = matching_files[1]
            ending_R1= "_S" + re.search(r"_S(\d+)", matching_files[0].split("/")[-1]).group(1) + re.split(r"_S\d+", matching_files[0])[1]
            ending_R2= "_S" + re.search(r"_S(\d+)", matching_files[1].split("/")[-1]).group(1) + re.split(r"_S\d+", matching_files[1])[1]
            metadata_2020_washed_amplicon_full.loc[index, "current_seq_sample_ID"] = re.split(r"_S\d+", matching_files[0].split("/")[-1])[0]
            metadata_2020_washed_amplicon_full.loc[index, "corrected_forward_fq"] = correct_seq_sample_ID + ending_R1
            metadata_2020_washed_amplicon_full.loc[index, "corrected_reverse_fq"] = correct_seq_sample_ID + ending_R2

        else:
            prefix2 = prefix.replace("washed-1-", "").replace("-", "_")
            matching_files2 = list_files_with_prefix(folder_path, prefix2)

            if len(matching_files2) == 2:
                metadata_2020_washed_amplicon_full.loc[index, "forward_fq"] = matching_files2[0]
                metadata_2020_washed_amplicon_full.loc[index, "reverse_fq"] = matching_files2[1]
                ending_R1= "_S" + re.search(r"_S(\d+)", matching_files2[0].split("/")[-1]).group(1) + re.split(r"_S\d+", matching_files2[0])[1]
                ending_R2= "_S" + re.search(r"_S(\d+)", matching_files2[1].split("/")[-1]).group(1) + re.split(r"_S\d+", matching_files2[1])[1]
                metadata_2020_washed_amplicon_full.loc[index, "current_seq_sample_ID"] = re.split(r"_S\d+", matching_files2[0].split("/")[-1])[0]
                metadata_2020_washed_amplicon_full.loc[index, "corrected_forward_fq"] = correct_seq_sample_ID + ending_R1
                metadata_2020_washed_amplicon_full.loc[index, "corrected_reverse_fq"] = correct_seq_sample_ID + ending_R2
            
        

    except Exception as e:
        # Print the error and the problematic row
        print(f"Error occurred for row {index}: {row}")
        print(f"Error details: {e}")
    
    # Print the percentage of completion
    percent_complete = (n + 1) / total_rows * 100
    print(f"Processing... {percent_complete:.2f}% complete", end="\r")
    n=n+1

# After the loop, print 100% to indicate completion
print("\nProcessing complete!")
metadata_2020_washed_amplicon_full
```

```python
# Define a list of columns to be moved to the end
columns_to_move = ["current_seq_sample_ID", "correct_seq_sample_ID", "suggested_correct_seq_sample_ID"]

# Reorder the DataFrame so that the specified columns are moved to the end
metadata_2020_washed_amplicon_full = metadata_2020_washed_amplicon_full[
    [col for col in metadata_2020_washed_amplicon_full.columns if col not in columns_to_move] + columns_to_move
]

# Extract only the file names from the 'forward_fq' column (removes the path)
metadata_2020_washed_amplicon_full['forward_fq'] = metadata_2020_washed_amplicon_full['forward_fq'].str.split("/").str[-1]

# Extract only the file names from the 'reverse_fq' column (removes the path)
metadata_2020_washed_amplicon_full['reverse_fq'] = metadata_2020_washed_amplicon_full['reverse_fq'].str.split("/").str[-1]

# Create a new column "SUGGESTED_EQUALS_CORRECT" that is True if the suggested correct seq_sample_ID equals the correct seq_sample_ID
metadata_2020_washed_amplicon_full["SUGGESTED_EQUALS_CORRECT"] = metadata_2020_washed_amplicon_full["suggested_correct_seq_sample_ID"] == metadata_2020_washed_amplicon_full["correct_seq_sample_ID"]

# Create a new column "ALREADY_CORRECT" that is True if the suggested correct seq_sample_ID equals the current seq_sample_ID
metadata_2020_washed_amplicon_full["ALREADY_CORRECT"] = metadata_2020_washed_amplicon_full["suggested_correct_seq_sample_ID"] == metadata_2020_washed_amplicon_full["current_seq_sample_ID"]

# Sort the DataFrame by "SUGGESTED_EQUALS_CORRECT" and "ALREADY_CORRECT" columns
# The rows with "True" for "SUGGESTED_EQUALS_CORRECT" and "ALREADY_CORRECT" will come first
metadata_2020_washed_amplicon_full = metadata_2020_washed_amplicon_full.sort_values(["SUGGESTED_EQUALS_CORRECT", "ALREADY_CORRECT"])

# Save the resulting DataFrame to a CSV file named 'test_metadata_15_11_2024.csv'
metadata_2020_washed_amplicon_full
```

```python
import re

metadata_all = pd.concat(
    {
        "2020_washed_amplicon": metadata_2020_washed_amplicon_full_SP,
        "2021_crushed_amplicon": metadata_2021_crushed_amplicon_full_SP,
        "2022_crushed_amplicon": metadata_2022_crushed_amplicon_full_SP,
        "2022_washed_shotgun": metadata_2022_washed_shotgun_full_SP,
        "2022_washed_amplicon": metadata_2022_washed_amplicon_full_SP,
        "2022_washed_shotgun_amplicon": metadata_2022_washed_shotgun_amplicon_full_SP,
        "2023_crushed_amplicon_16S": metadata_2023_crushed_amplicon_16S_full_SP,
        "2023_crushed_amplicon_ITS": metadata_2023_crushed_amplicon_ITS_full_SP,
        "control_crushed_amplicon": metadata_control_crushed_amplicon_full_SP,
        "2023_washed_shotgun_amplicon": metadata_2023_washed_shotgun_amplicon_full_SP,
        "2023_washed_shotgun": metadata_2023_washed_shotgun_full_SP,
    },
    names=["source"]
).reset_index(level=0).reset_index(drop=True)

metadata_all["is_rerun"] = False
metadata_all["sequencing_run"] = metadata_all["run_ID"]

rerun_samples = pd.read_csv("../EXCEL_FILES/BRLUV_DR8PG_rerun_samples.csv").copy()

rerun_to_original_run = {
    "seq260520_BRLUV": "seq240412_BRLUV",
    "seq260612_DR8PG": "seq240710_DR8PG",
}

accepted_rerun_only_templates = {
    ("seq260612_DR8PG", "T-130623-ITS-c-10C-1"): ("seq240710_DR8PG", "T-2023-ITS-c-P2-neg"),
}

unknown_rerun_runs = sorted(set(rerun_samples["run_ID"]) - set(rerun_to_original_run))
if unknown_rerun_runs:
    raise ValueError(f"Missing rerun_to_original_run entries for: {unknown_rerun_runs}")

def _current_seq_sample_id(fastq_name):
    return re.split(r"_S\d+", str(fastq_name).split("/")[-1], maxsplit=1)[0]

def _fastq_suffix(fastq_name):
    match = re.search(r"(_S\d+.*)$", str(fastq_name).split("/")[-1])
    return match.group(1) if match else ""

rerun_metadata_rows = []
missing_rerun_templates = []
ambiguous_rerun_templates = []

for rerun in rerun_samples.itertuples(index=False):
    rerun_run = str(rerun.run_ID)
    seq_sample_id = str(rerun.seq_sample_ID)
    original_run = rerun_to_original_run[rerun_run]
    template_run, template_seq_sample_id = accepted_rerun_only_templates.get(
        (rerun_run, seq_sample_id),
        (original_run, seq_sample_id),
    )

    template = metadata_all[
        (metadata_all["run_ID"].astype(str) == template_run)
        & (metadata_all["seq_sample_ID"].astype(str) == template_seq_sample_id)
    ]

    if template.empty:
        missing_rerun_templates.append((rerun_run, seq_sample_id, template_run, template_seq_sample_id))
        continue
    if len(template) > 1:
        ambiguous_rerun_templates.append((rerun_run, seq_sample_id, template_run, template_seq_sample_id, len(template)))
        continue

    rerun_row = template.iloc[0].copy()
    rerun_row["run_ID"] = rerun_run
    rerun_row["sequencing_run"] = rerun_run
    rerun_row["is_rerun"] = True
    rerun_row["seq_sample_ID"] = seq_sample_id
    rerun_row["forward_fq"] = rerun.forward_fq
    rerun_row["reverse_fq"] = rerun.reverse_fq

    if "current_seq_sample_ID" in rerun_row.index:
        rerun_row["current_seq_sample_ID"] = _current_seq_sample_id(rerun.forward_fq)

    corrected_id = seq_sample_id
    if "suggested_correct_seq_sample_ID" in rerun_row.index and pd.notna(rerun_row["suggested_correct_seq_sample_ID"]):
        corrected_id = str(rerun_row["suggested_correct_seq_sample_ID"])

    if "corrected_forward_fq" in rerun_row.index:
        rerun_row["corrected_forward_fq"] = corrected_id + _fastq_suffix(rerun.forward_fq)
    if "corrected_reverse_fq" in rerun_row.index:
        rerun_row["corrected_reverse_fq"] = corrected_id + _fastq_suffix(rerun.reverse_fq)

    rerun_metadata_rows.append(rerun_row)

if missing_rerun_templates:
    raise ValueError(f"Missing source metadata rows for reruns: {missing_rerun_templates[:10]}")
if ambiguous_rerun_templates:
    raise ValueError(f"Ambiguous source metadata rows for reruns: {ambiguous_rerun_templates[:10]}")

rerun_metadata = pd.DataFrame(rerun_metadata_rows)

if "dir_name" in rerun_metadata.columns:
    rerun_metadata["dir_name"] = rerun_metadata["run_ID"]
if "full_path" in rerun_metadata.columns:
    run_path_info = df_folder_structure.drop_duplicates("dir_name").set_index("dir_name")
    rerun_metadata["full_path"] = rerun_metadata["run_ID"].map(run_path_info["full_path"])
if "Unnamed: 0" in rerun_metadata.columns and "Unnamed: 0" in df_folder_structure.columns:
    run_path_info = df_folder_structure.drop_duplicates("dir_name").set_index("dir_name")
    rerun_metadata["Unnamed: 0"] = rerun_metadata["run_ID"].map(run_path_info["Unnamed: 0"])

if len(rerun_metadata) != len(rerun_samples):
    raise ValueError(f"Expected {len(rerun_samples)} rerun rows, got {len(rerun_metadata)}")

metadata_all = pd.concat([metadata_all, rerun_metadata], ignore_index=True, sort=False)
metadata_all.to_csv(metadata_output_dir / "metadata_all.csv", index=False)
```
